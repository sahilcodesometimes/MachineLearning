```python
import numpy as np
from collections import Counter
```

```python
X_train = np.array([
    [1,2],
    [2,1],
    [3,3],
    [4,4]
])

y_train = np.array([0,0,1,1])
```

```python
X_test = np.array([
    [3,2],
    [5,4]
])
y_test = np.array([1,1])
```

```python
def knn_predict(X_train,X_test,y_train,k=3):
    predictions = []
    for new_points in X_test:
        distance = np.sqrt(np.sum( X_train - new_points ) ** 2)
        nearest_indices = np.argsort(distance)[:k]
        nearest_labels = y_train[nearest_indices]

        predication = Counter(nearest_labels).most_common(1)[0][0]
        predictions.append(prediction)
    return np.array(predictions)
```

```python
predictions = knn_predict(X_train, X_test,y_train, k = 3)
print(predictions)
```
