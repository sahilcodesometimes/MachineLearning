```python
maths = 85
python = 92
statistics = 74
```

```python
total_marks = maths + python + statistics
avg_marks = total_marks / 3
print(f"Total Marks : {total_marks}")
print(f"Average Marks : {round(avg_marks,2)}")
```

```python
avg_greater_then_90 = (avg_marks >= 90)
all_sub_marks_greater_70 = (maths and python and statistics >= 70)
print(f"is average marks greater then 90 : {avg_greater_then_90}")
print(f"is all subjects marks greater then 70 : {all_sub_marks_greater_70}")
```
